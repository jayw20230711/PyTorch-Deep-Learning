```python
import numpy as np
import matplotlib.pyplot as plt

# Define the tanh function
def tanh(w, b, x):
    return np.tanh(w * x + b)

# Generate x values
x = np.linspace(-10, 10, 1000)  # Range of x values
w = 1  # Weight
b = 0  # Bias
y = tanh(w, b, x)  # Compute tanh(wx + b)

# Plot the curve
plt.plot(x, y, label=f"tanh({w}x + {b})")
plt.axhline(0, color="black", linestyle="--", linewidth=0.5)  # Horizontal line at y=0
plt.axvline(0, color="black", linestyle="--", linewidth=0.5)  # Vertical line at x=0
plt.xlabel("wx + b")
plt.ylabel("tanh(wx + b)")
plt.title("S-Shaped Curve of tanh(wx + b)")
plt.legend()
plt.grid()
plt.show()
```
