```python
import torch

# Create a tensor of zeros
zeros_tensor = torch.zeros(2, 3)  # 2x3 tensor filled with zeros
zeros_tensor
```

```python
# Create a tensor of ones
ones_tensor = torch.ones(2, 3)  # 2x3 tensor filled with ones
ones_tensor
```

```python
# Create a tensor with random values
random_tensor = torch.rand(2, 3)  # 2x3 tensor with values uniformly sampled from [0, 1)
random_tensor
```

```python
# Create a tensor with values sampled from a normal distribution
normal_tensor = torch.randn(2, 3)  # 2x3 tensor with values from a standard normal distribution
normal_tensor
```
